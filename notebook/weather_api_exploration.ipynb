# Live Weather Data Platform: Api Extraction


## Configuration

```python
API_URL = "https://api.open-meteo.com/v1/forecast"

LOCATION = {
    "name": "Brasilia",
    "latitude": -15.7939,
    "longitude": -47.8828,
    "timezone": "America/Sao_Paulo",
}

FORECAST_DAYS = 7

HOURLY_VARIABLES = [
    "temperature_2m",
    "relative_humidity_2m",
    "apparent_temperature",
    "precipitation",
    "rain",
    "weather_code",
    "wind_speed_10m",
    "wind_direction_10m",
    "wind_gusts_10m",
    "surface_pressure",
]

REQUEST_PARAMS = {
    "latitude": LOCATION["latitude"],
    "longitude": LOCATION["longitude"],
    "hourly": HOURLY_VARIABLES,
    "forecast_days": FORECAST_DAYS,
    "timezone": LOCATION["timezone"],
}

print("API endpoint:", API_URL)
print("Location:", LOCATION["name"])
print("Forecast days:", FORECAST_DAYS)
print("Hourly variables:", len(HOURLY_VARIABLES))
```

## Extract Weather Data

```python
def fetch_weather_data(): 
    """Fetch weather forecast data from the Open-Meteo API."""

    try:
        response = requests.get(
            API_URL,
            params=REQUEST_PARAMS,
            timeout=30,
        )

        response.raise_for_status()

        return response.json()

    except requests.exceptions.RequestException as error:
        print(f"Weather API request failed: {error}")
        raise
```

```python
# Save the transformation logic as a reuseable function

def transform_weather_data(raw_response): 
    """Transform a raw Open-Meteo response into a clean hourly DataFrame."""

    if not isinstance(raw_response, dict):
        raise TypeError("Expected the API response to be a dictionary.")

    if "hourly" not in raw_response:
        raise ValueError("API response is missing the 'hourly' section.")

    hourly_data = raw_response["hourly"]

    if "time" not in hourly_data:
        raise ValueError("Hourly data is missing timestamps.")

    weather_df = pd.DataFrame(hourly_data)

    if weather_df.empty:
        raise ValueError("The API returned no hourly weather records.")

    weather_df["time"] = pd.to_datetime(
        weather_df["time"],
        errors="raise",
    )

    if weather_df["time"].isna().any():
        raise ValueError("Missing timestamps detected.")

    if weather_df["time"].duplicated().any():
        raise ValueError("Duplicate timestamps detected.")

    if not weather_df["time"].is_monotonic_increasing:
        raise ValueError("Timestamps are not in chronological order.")

    return weather_df
```

```python
# Checking Data Quality Independently

def validate_weather_data(weather_df):
    """Validate the structure and basic quality of hourly weather data."""

    required_columns = [
        "time",
        "temperature_2m",
        "relative_humidity_2m",
        "precipitation",
        "wind_speed_10m",
    ]

    missing_columns = [
        column
        for column in required_columns
        if column not in weather_df.columns
    ]

    if missing_columns:
        raise ValueError(f"Missing required columns: {missing_columns}")

    if weather_df.empty:
        raise ValueError("Weather DataFrame is empty.")

    if weather_df["time"].isna().any():
        raise ValueError("Missing timestamps detected.")

    if weather_df["time"].duplicated().any():
        raise ValueError("Duplicate timestamps detected.")

    if not weather_df["time"].is_monotonic_increasing:
        raise ValueError("Timestamps are not in chronological order.")

    if not weather_df["temperature_2m"].between(-90, 60).all():
        raise ValueError("Temperature outside sanity limits.")

    if not weather_df["relative_humidity_2m"].between(0, 100).all():
        raise ValueError("Humidity outside valid limits.")

    if not weather_df["precipitation"].ge(0).all():
        raise ValueError("Negative precipitation detected.")

    if not weather_df["wind_speed_10m"].ge(0).all():
        raise ValueError("Negative wind speed detected.")

    return True
```
